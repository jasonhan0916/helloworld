 ## Temporal distribution

```python
# Parse BEGIN_DATE safely even if formats differ or cells are blank
hail['BEGIN_DATE'] = pd.to_datetime(hail['BEGIN_DATE'], errors='coerce', format='mixed')

# OPTIONAL: if two-digit years like "55" were parsed as 2055, push them back a century
hail.loc[hail['BEGIN_DATE'].dt.year > 2025, 'BEGIN_DATE'] = \
    hail.loc[hail['BEGIN_DATE'].dt.year > 2025, 'BEGIN_DATE'] - pd.DateOffset(years=100)

# Now add Year/Month columns
hail['Year']  = hail['BEGIN_DATE'].dt.year
hail['Month'] = hail['BEGIN_DATE'].dt.month

# Counts
hail_by_year   = hail['Year'].value_counts().sort_index()
hail_by_month  = hail['Month'].value_counts().sort_index()

print("Hail events per year:")
display(hail_by_year)

print("\nHail events per month:")
display(hail_by_month)
```

```python
# Create a line plot for hail events per year
plt.figure(figsize=(12, 6))
sns.lineplot(x=hail_by_year.index, y=hail_by_year.values)
plt.title('Hail Events Per Year (Line Plot)')
plt.xlabel('Year')
plt.ylabel('Number of Hail Events')
plt.show()

# Create a bar plot for hail events per month
plt.figure(figsize=(10, 6))
sns.barplot(x=hail_by_month.index, y=hail_by_month.values)
plt.title('Hail Events Per Month (Bar Plot)')
plt.xlabel('Month')
plt.ylabel('Number of Hail Events')
plt.xticks(range(0, 12), ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec'])
plt.show()
```

```python
# Identify the year with the highest frequency
most_frequent_year = hail_by_year.idxmax()
highest_year_count = hail_by_year.max()
print(f"Year with the highest frequency of hail events: {most_frequent_year} ({highest_year_count} events)")

# Identify the month with the highest frequency
most_frequent_month = hail_by_month.idxmax()
highest_month_count = hail_by_month.max()
print(f"Month with the highest frequency of hail events: {most_frequent_month} ({highest_month_count} events)")
```

There are lots of hails in the late 1990s. During 2000s, there are up and downs in hail number. After 2010, hail events decrease.

## Spatial distribution

```python
# Count hail events per state
hail_by_state = hail['STATE_ABBR'].value_counts()

# Create a bar plot for hail events per state
plt.figure(figsize=(12, 6))
sns.barplot(x=hail_by_state.index, y=hail_by_state.values)
plt.title('Hail Events Per State')
plt.xlabel('State Abbreviation')
plt.ylabel('Number of Hail Events')
plt.xticks(rotation=45, ha='right') # Rotate x-axis labels for better readability
plt.tight_layout() # Adjust layout to prevent labels overlapping
plt.show()
```

States like North Carolina, Georgia, Mississipi, Atlanta have a lot of hail events. States like Maine, New Hampshire, and DC don't have much.

##Data cleaning

```python
# Convert MAGNITUDE to numeric, coercing errors
hail['MAGNITUDE'] = pd.to_numeric(hail['MAGNITUDE'], errors='coerce')

# Convert inches to centimeters (1 inch = 2.54 cm)
hail['MAGNITUDE_CM'] = hail['MAGNITUDE'] * 2.54

print("Hail data with magnitude in centimeters:")
display(hail[['MAGNITUDE', 'MAGNITUDE_CM']].head())
```

```python
state_map = {
    'AL': 'Alabama', 'GA': 'Georgia', 'FL': 'Florida', 'LA': 'Louisiana',
    'ME': 'Maine', 'MS': 'Mississippi', 'NC': 'North Carolina',
    'NH': 'New Hampshire', 'SC': 'South Carolina', 'WV': 'West Virginia',
    'DC': 'District of Columbia'
}

hail['STATE_FULL_NAME'] = hail['STATE_ABBR'].map(state_map)

print("Hail data with full state names:")
display(hail[['STATE_ABBR', 'STATE_FULL_NAME']])
```
