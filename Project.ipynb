## 4. Feature Engineering


We create a new feature: Vaccinated Percentage.

This represents the tip as a percentage of Vaccinated Population. It is often more insightful to look at Vaccinated percentage rather than the raw Vaccination Population, as it normalizes the population vaccinated relative to the whole population.

We can compute this new column as: **vaccinated_percentage** = [ vac_pop / (vac_pop+ population unvaccinated) ]* 100.

We'll add this column to the DataFrame:



```python
# Create a new feature for vaccinated percentage of the total population
covid['vaccinated_percentage'] = (covid['Vac_Pop']/(covid['Vac_Pop']+covid['Population Unvaccinated']))*100

# Display the first five rows to see the new feature
print(covid.head())
```

We calculate the vaccinated percentage by dividing the population vaccinated by total population and multiplying by 100.

The result is stored in a new column **vaccinated_percentage**.

After adding this feature, we print the first two rows again to verify that the **vaccinated_percentage** column appears with the expected values.

We create a new feature: **Boosted Percentage**.

This represents the population boosted divided by the population vaccinated.

Analyzing Boosted Percentage can be more insightful than looking at population vaccinated only, as it normalizes the percentage of accepting another booster shot per vaccinated individual, making comparisons across different population sizes and age groups more meaningful.

## 7. Statistical Analysis

To further quantify the effects of age group on vaccinated percentage, we perform a linear regression analysis.

```python
import statsmodels.api as sm

# 1) Make sure we only have numeric columns in X
X_multi = pd.get_dummies(covid['Age Group'], drop_first=True)

# 2) Add a constant intercept
X_multi = sm.add_constant(X_multi)

# 3) Convert to float (just in case)
X_multi = X_multi.astype(float)

# 4) Our target y is price_per_carat, ensure numeric
y = covid['vaccinated_percentage'].astype(float)

# 5) Fit OLS
model = sm.OLS(y, X_multi).fit()
print(model.summary())
```

###Interpretation of the Regression Results


This linear model (OLS) explains a large portion of the variance in vaccinated_percentage (R‐squared ≈ 0.963), indicating that age group alone is highly predictive of vaccination levels in this dataset. The overall model is statistically significant (F‐statistic = 4344, p < 0.001).

The baseline (reference) category is “12–17,” with an intercept of about 71.67% vaccinated.

The null hypothesis is that the vaccination percentage for this age group does not differ from the reference group (12–17); The alternative hypothesis is that the vaccination percentage for this age group does differ from the reference group. A two–sided t-test is reported for every coefficient, with significance assessed at the 5 % level (α = 0.05).

A positive coefficient means that age group has a higher vaccination percentage than the reference, while a negative coefficient indicates a lower percentage:

5–11: –24.29% (p < 0.001) — significantly lower than 12–17.(Reject)

18–29: –0.32% (p = 0.052) — borderline significance, near the reference level. (Fail to reject)

30–49: –1.10% (p = 0.001) — modestly but significantly lower. (Reject)

50–64: +8.64% (p < 0.001) — significantly higher. (Reject)

65–79: –0.99% (p = 0.025) — also slightly lower. (Reject)

80+: –14.48% (p < 0.001) — significantly lower. (Reject)


In practical terms, children ages 5–11 show the largest drop in vaccination percentage compared to adolescents, while adults 50–64 have the highest relative increase. Other age groups exhibit smaller shifts around the baseline. These findings suggest that age group strongly correlates with vaccination coverage, though factors beyond age may also play a role.
